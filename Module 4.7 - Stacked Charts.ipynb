### Task 1: Stacked Chart for Composition of Job Satisfaction Across Age Groups


##### 1. Stacked Chart of Median `JobSatPoints_6` and `JobSatPoints_7` for Different Age Groups


Visualize the composition of job satisfaction scores (`JobSatPoints_6` and `JobSatPoints_7`) across various age groups. This will help in understanding the breakdown of satisfaction levels across different demographics.



```python
##Write your code here
df_jspa67 = df.copy()
df_jspa67 = df_jspa67.groupby('Age')[['JobSatPoints_6' , 'JobSatPoints_7']].median().reset_index()
df_jspa67.plot(kind = 'bar' , stacked = True , x = 'Age' , y = ['JobSatPoints_6', 'JobSatPoints_7'] , figsize = (8 , 5))
plt.xlabel('Age Group')
plt.ylabel('Median Job Satisfaction Points')
plt.title('Median Job Satisfaction by Age Group')
plt.show()
```

##### Stacked Chart of `JobSatPoints_6` and `JobSatPoints_7` for Employment Status


Create a stacked chart to compare job satisfaction (`JobSatPoints_6` and `JobSatPoints_7`) across different employment statuses. This will show how satisfaction varies by employment type.


```python
##Write your code here
#df['Employment'].value_counts()
df_jspe67 = df.copy()
df_jspe67['EmploymentGrouped'] = df_jspe67['Employment'].str.split(';').str[0].str.strip()
#df_jspe67['EmploymentGrouped'].value_counts()
df_jspe67 = df_jspe67.groupby('EmploymentGrouped')[['JobSatPoints_6' , 'JobSatPoints_7']].median().reset_index()
#df_jspe67
df_jspe67.plot(kind = 'bar' , stacked = True , x = 'EmploymentGrouped' , y = ['JobSatPoints_6', 'JobSatPoints_7'] , figsize = (8 , 5))
plt.xlabel('Employment Group')
plt.ylabel('Median Job Satisfaction Points')
plt.title('Median Job Satisfaction by Employment Group')
plt.show()
```

### Task 2: Stacked Chart for Compensation and Job Satisfaction by Age Group


##### This stacked chart visualizes the composition of compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`) specifically for respondents aged 30-35.


```python
##Write your code here
#df['Age'].value_counts()
df_ccj6 = df.copy()
q1_compy = df_ccj6['ConvertedCompYearly'].describe()['25%']
q3_compy = df_ccj6['ConvertedCompYearly'].describe()['75%']
iqr_compy = q3_compy - q1_compy
Lbound_compy = q1_compy - 1.5 * iqr_compy
Ubound_compy = q3_compy + 1.5 * iqr_compy
df_ccj6['Compy'] = df_ccj6['ConvertedCompYearly'].where((df_ccj6['ConvertedCompYearly'] >= Lbound_compy) & (df_ccj6['ConvertedCompYearly'] <= Ubound_compy))
import numpy as np
age_ranges = {
    'Under 18 years old': (0, 17),
    '18-24 years old': (18, 24),
    '25-34 years old': (25, 34),
    '35-44 years old': (35, 44),
    '45-54 years old': (45, 54),
    '55-64 years old': (55, 64),
    '65 years or older': (65, 90),
    'Prefer not to say': (0 , 90)
}

def random_age(age_grp):
    if age_grp in age_ranges:
        low , high = age_ranges[age_grp]
        return np.random.randint(low , high + 1)
    return np.nan

df_ccj6['AgeNumber'] = df_ccj6['Age'].apply(random_age)
age30_35 = df_ccj6[(df_ccj6['AgeNumber'] >= 30) & (df_ccj6['AgeNumber'] <= 35)]
df_ccj6 = age30_35.groupby('AgeNumber')[['JobSatPoints_6' , 'Compy']].median().reset_index()
df_ccj6.plot(kind = 'bar' , stacked = True , x = 'AgeNumber' , y = ['JobSatPoints_6', 'Compy'] , figsize = (8 , 5))
plt.xlabel('Age')
plt.ylabel('Median Job Satisfaction Points and Compensation')
plt.title('Median Job Satisfaction and Compensation by Age (30-35)')
plt.show()
```

##### Stacked Chart of Median Compensation and Job Satisfaction Across Age Group


Compare the median compensation and job satisfaction metrics across different age groups. This helps visualize how compensation and satisfaction levels differ by age.


```python
##Write your code here
df_ccj = df.copy()
q1_compy = df_ccj['ConvertedCompYearly'].describe()['25%']
q3_compy = df_ccj['ConvertedCompYearly'].describe()['75%']
iqr_compy = q3_compy - q1_compy
Lbound_compy = q1_compy - 1.5 * iqr_compy
Ubound_compy = q3_compy + 1.5 * iqr_compy
df_ccj['Compy'] = df_ccj['ConvertedCompYearly'].where((df_ccj['ConvertedCompYearly'] >= Lbound_compy) & (df_ccj['ConvertedCompYearly'] <= Ubound_compy))
df_ccj = df_ccj.groupby('Age')[['JobSat' , 'Compy']].median().reset_index()
df_ccj.plot(kind = 'bar' , stacked = True , x = 'Age' , y = ['JobSat', 'Compy'] , figsize = (8 , 5))
plt.xlabel('Age')
plt.ylabel('Median Job Satisfaction Points and Compensation')
plt.title('Median Job Satisfaction and Compensation by Age group')
plt.show()
```

### Task 3: Comparing Data Using Stacked Charts


##### 1. Stacked Chart of Preferred Databases by Age Group




Visualize the top databases that respondents from different age groups wish to learn. Create a stacked chart to show the proportion of each database in each age group.


```python
##Write your code here
#df['DatabaseWantToWorkWith'].value_counts()
df_dbwa = df.copy()
df_dbwa['DatabaseWantToWorkWith'] = df_dbwa['DatabaseWantToWorkWith'].str.split(';')
df_dbwa = df_dbwa.explode('DatabaseWantToWorkWith')
df_dbwa = df_dbwa.reset_index(drop=True)
df_dbwa['DatabaseWantToWorkWith'] = df_dbwa['DatabaseWantToWorkWith'].str.strip()
df_dbwa['Age'] = df_dbwa['Age'].str.strip()
df_dbwa = df_dbwa.dropna(subset=['Age', 'DatabaseWantToWorkWith'])
df_dbwa5 = df_dbwa['DatabaseWantToWorkWith'].value_counts().head(5).index
df_dbwa = df_dbwa[df_dbwa['DatabaseWantToWorkWith'].isin(df_dbwa5)]
dbwa = pd.crosstab(df_dbwa['Age'] , df_dbwa['DatabaseWantToWorkWith'])
dbwa_prop = dbwa.div(dbwa.sum(axis=1), axis=0)
dbwa_prop.plot(kind = 'bar' , stacked = True , figsize = (10 , 5))
plt.xlabel('Age Group')
plt.ylabel('Proportion of Database Preferences')
plt.title('Top 5 Preferred Databases by Age Group')
plt.legend(loc='upper left', bbox_to_anchor=(1.02, 1), borderaxespad=0.)
plt.show()
```

##### 2. Stacked Chart of Employment Type by Job Satisfaction


Analyze the distribution of employment types within each job satisfaction level using a stacked chart. This will provide insights into how employment types are distributed across various satisfaction ratings.


```python
##Write your code here
df_jse = df.copy()
def clean_employ(value):
    if pd.isna(value):
        return 'Unknown'

    value = value.lower()

    if 'employed, full-time' in value:
        return 'Employed, full-time'

    elif 'employed, part-time' in value:
        return 'Employed, part-time'

    elif 'self-employed' in value or 'freelancer' in value:
        return 'Self-employed'
        
    elif 'student' in value:
        return 'Student'

    elif 'not employed' in value:
        return 'Unemployed'

df_jse['EmploymentGrouped'] = df_jse['Employment'].apply(clean_employ)
#df_jse['EmploymentGrouped'].value_counts()
jse = pd.crosstab(df_jse['JobSat'] , df_jse['EmploymentGrouped'])
jse.plot(kind = 'bar' , stacked = True , figsize = (8 , 5))
plt.xlabel('Job Satisfaction rating')
plt.ylabel('Frequency')
plt.title('Proportion of Employment type by Job Satisfaction')
plt.legend(loc='upper left', bbox_to_anchor=(1.02, 1), borderaxespad=0.)
plt.show()
```

### Task 4: Exploring Technology Preferences Using Stacked Charts


##### 1. Stacked Chart for Preferred Programming Languages by Age Group


Analyze how programming language preferences (`LanguageAdmired`) vary across age groups.


```python
##Write your code here
df_laa = df.copy()
df_laa['LanguageAdmired'] = df_laa['LanguageAdmired'].str.split(';')
df_laa = df_laa.explode('LanguageAdmired')
df_laa = df_laa.reset_index(drop=True)
df_laa['LanguageAdmired'] = df_laa['LanguageAdmired'].str.strip()
df_laa['Age'] = df_laa['Age'].str.strip()
df_laa = df_laa.dropna(subset=['Age', 'LanguageAdmired'])
df_laa5 = df_laa['LanguageAdmired'].value_counts().head(5).index
df_laa = df_laa[df_laa['LanguageAdmired'].isin(df_laa5)]
laa = pd.crosstab(df_laa['Age'],df_laa['LanguageAdmired'])
laa_prop = laa.div(laa.sum(axis=1), axis=0)
laa_prop.plot(kind = 'bar' , stacked = True , figsize = (10 , 5))
plt.xlabel('Age Group')
plt.ylabel('Proportion of Admired Languages')
plt.title('Top 5 Admired Programming Languages by Age Group')
plt.legend(loc='upper left', bbox_to_anchor=(1.02, 1), borderaxespad=0.)
plt.show()
```

##### 2. Stacked Chart for Technology Adoption by Employment Type


Explore how admired platforms (`PlatformAdmired`) differ across employment types (e.g., full-time, freelance)


```python
##Write your code here
df_pae = df.copy()
df_pae['PlatformAdmired'] = df_pae['PlatformAdmired'].str.split(';')
df_pae = df_pae.explode('PlatformAdmired')
df_pae = df_pae.reset_index(drop=True)
df_pae['PlatformAdmired'] = df_pae['PlatformAdmired'].str.strip()
#df_pae['PlatformAdmired'].value_counts()
df_pae5 = df_pae['PlatformAdmired'].value_counts().head(5).index
df_pae = df_pae[df_pae['PlatformAdmired'].isin(df_pae5)]
df_pae['EmploymentGrouped'] = df_pae['Employment'].apply(clean_employ)
#df_pae['EmploymentGrouped'].value_counts()
pae = pd.crosstab(df_pae['EmploymentGrouped'] , df_pae['PlatformAdmired'])
pae_prop = pae.div(pae.sum(axis=1), axis=0)
pae_prop.plot(kind = 'bar' , stacked = True , figsize = (10 , 5))
plt.xlabel('Employment Group')
plt.ylabel('Proportion of platforms admired')
plt.title('Top 5 Admired platforms by Employment Group')
plt.legend(loc='upper left', bbox_to_anchor=(1.02, 1), borderaxespad=0.)
plt.show()
```
