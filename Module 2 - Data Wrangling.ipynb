### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
#Inspect entire country column
df['Country'].value_counts()
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Looks like there are multiple different entries for north/south korea under different names
df['Country'][df['Country'].str.contains('korea' , case = False , na = False)].unique()
```

```python
#standardize formatting
df['Country'] = df['Country'].str.strip().str.lower()

#replace inconsistent names for south korea
df['Country'] = df['Country'].replace({'republic of korea' : 'south korea' , "democratic people's republic of korea" : 'north korea'})

#check it worked
df['Country'][df['Country'].str.contains('korea' , case = False , na = False)].unique()
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here
employment_encoded = pd.get_dummies(df['Employment'] , dtype = int)
employment_encoded.head()
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
df.isna().sum().sort_values(ascending = False)
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
print('missing values:' , df['ConvertedCompYearly'].isna().sum())
ConvertedCompYearly_median = df['ConvertedCompYearly'].median()
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(ConvertedCompYearly_median)
df['ConvertedCompYearly'].head()
print('missing values filled with median:' , df['ConvertedCompYearly'].isna().sum())
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
print('missing values:' , df['RemoteWork'].isna().sum())
remotework_mode = df['RemoteWork'].mode()[0]
df['RemoteWork'] = df['RemoteWork'].fillna(remotework_mode)
print('missing values filled with mode:' , df['RemoteWork'].isna().sum())
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
min_compy = df['ConvertedCompYearly'].min()
max_compy = df['ConvertedCompYearly'].max()
df['ConvertedCompYearly_MinMax'] = (df['ConvertedCompYearly'] - min_compy) / (max_compy - min_compy)
df['ConvertedCompYearly_MinMax'].head()
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import numpy as np
df['ConvertedCompYearly_log'] = np.log(df['ConvertedCompYearly'])
df['ConvertedCompYearly_log'].head()
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
##column has numbers stored as strings, and inconsistent values like 'less than 1 year' so we need to clean the column first.
df['YearsCodePro'].unique()
```

```python
#replace inconsistent values
df['YearsCodePro'] = df['YearsCodePro'].replace({'Less than 1 year' : 0.5 , 'More than 50 years' : 51})

#convert to numeric values
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'] , errors = 'coerce')

#fill in missing values
yearscodepro_median = df['YearsCodePro'].median()
df['YearsCodePro'] = df['YearsCodePro'].fillna(yearscodepro_median)
print('missing values:' , df['YearsCodePro'].isna().sum())
```

```python
#create 'ExperienceLevel' column by turning years coded into 3 bins representing experience levels
bins = [0 , 3 , 10 , 100]
labels = ['entry-level' , 'mid-level' , 'senior-level']
df['ExperienceLevel'] = pd.cut(df['YearsCodePro'] , bins = bins , labels = labels)

df['ExperienceLevel'].value_counts()
```
