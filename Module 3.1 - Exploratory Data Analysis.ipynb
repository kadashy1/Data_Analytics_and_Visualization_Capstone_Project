#### Step 2: Load and Preview the Dataset

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
##check for % of missing values in columns
employ_jobsat_remote = ['Employment' , 'JobSat' , 'RemoteWork']
df[employ_jobsat_remote].isna().mean() * 100
```

```python
#missing values in remaining columns will be filled using distribution based imputation, as 55% and 16% are too high to simply remove from the data set.
#using the mode to fill in such large portions of missing data would skew the data too much.
import numpy as np
#create a new copy of data set to modify
df_clean = df.copy()
#fill missing values in remotework
proportion_r = df['RemoteWork'].value_counts(normalize = True)
missing_r = df['RemoteWork'].isna()
df_clean.loc[missing_r , 'RemoteWork'] = np.random.choice(proportion_r.index , size = missing_r.sum() , p = proportion_r.values)
#check if it worked
print('missing values in original df:' , df['RemoteWork'].isna().sum())
print('missing values in df_clean:' , df_clean['RemoteWork'].isna().sum())
```

```python
#do the same thing for JobSat
proportion_j = df['JobSat'].value_counts(normalize = True)
missing_j = df['JobSat'].isna()
df_clean.loc[missing_j , 'JobSat'] = np.random.choice(proportion_j.index , size = missing_j.sum() , p = proportion_j.values)
#check it worked
print('missing values in original:' , df['JobSat'].isna().sum())
print('missing values in df_clean:' , df_clean['JobSat'].isna().sum())
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
##we will start by inspecting and cleaning the YearsCodePro column
df_clean['YearsCodePro'].unique()
```

```python
#replace inconsistent values
df_clean['YearsCodePro'] = df_clean['YearsCodePro'].replace({'Less than 1 year' : 0.5 , 'More than 50 years' : 51})

#convert the column into numeric values
df_clean['YearsCodePro'] = pd.to_numeric(df_clean['YearsCodePro'] , errors = 'coerce')

#and fill missing values
median_ypc = df_clean['YearsCodePro'].median()
df_clean['YearsCodePro'] = df_clean['YearsCodePro'].fillna(median_ypc)
print('missing values:' , df_clean['YearsCodePro'].isna().sum())
```

```python
#create 'ExperienceLevel' column to represent data from 'YearsCodePro' in bins
bins = [0 , 3 , 10 , 100]
labels = ['entry-level' , 'mid-level' , 'senior-level']
df_clean['ExperienceLevel'] = pd.cut(df_clean['YearsCodePro'] , bins = bins , labels = labels)
df_clean['ExperienceLevel'].value_counts()
```

```python
#calculate median 'JobSat' in 'ExperienceLevel'
median_je = df_clean.groupby('ExperienceLevel')['JobSat'].median()
print(median_je)
```

```python
#visualize relationship between the two columns
plt.figure(figsize = (10 , 5))

plt.bar(median_je.index.astype(str) , median_je.values)

plt.xlabel('Experience level')
plt.ylabel('Median job satisfaction')
plt.title('Median job satisfaction by experience level')
plt.show()
```

Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
##RemoteWork distribution
remoteworkcount = df_clean['RemoteWork'].value_counts()
plt.figure(figsize = (10 , 5))
plt.bar(remoteworkcount.index , remoteworkcount.values)
plt.xlabel('type of work')
plt.ylabel('Count')
plt.title('Distribution of remote work vs in person')
plt.show()
```

```python
#relationship between 'RemoteWork' and 'Employment'
#standardize labels in 'Employment'
def clean_employ(value):
    if pd.isna(value):
        return 'Unknown'

    value = value.lower()

    if 'employed, full-time' in value:
        return 'Employed, full-time'

    elif 'employed, part-time' in value:
        return 'Employed, part-time'

    elif 'self-employed' in value or 'freelancer' in value:
        return 'Self-employed'
        
    elif 'student' in value:
        return 'Student'

    elif 'not employed' in value:
        return 'Unemployed'

df_clean['EmploymentGrouped'] = df_clean['Employment'].apply(clean_employ)
df_clean['EmploymentGrouped'].value_counts()
```

```python
#create crosstab of employment and remotework
remote_counts = pd.crosstab(df_clean['EmploymentGrouped'] , df_clean['RemoteWork'])

#plot the relationship
remote_counts.plot(kind = 'bar' , figsize = (10 , 6))
plt.xlabel('Employment status')
plt.ylabel('Count')
plt.title('Remote work preference by employment type')
plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
##theres way too many values in each of these columns to visualize in one bar graph
print(df_clean['LanguageHaveWorkedWith'].value_counts())
print(df_clean['Country'].value_counts())
```

```python
#so we will just use the 5 most popular languages in the 5 countries with the most responses to the survey
top5_country = df_clean['Country'].value_counts().head(5).index
df_top = df_clean[df_clean['Country'].isin(top5_country)]

df_languagecountry = df_top.copy()
df_languagecountry['LanguageHaveWorkedWith'] = df_languagecountry['LanguageHaveWorkedWith'].str.split(';')
df_languagecountry = df_languagecountry.explode('LanguageHaveWorkedWith')
df_languagecountry = df_languagecountry.reset_index(drop = True)
df_languagecountry['LanguageHaveWorkedWith'] = df_languagecountry['LanguageHaveWorkedWith'].str.strip()

language5 = df_languagecountry['LanguageHaveWorkedWith'].value_counts().head(5).index
df_languagecountry = df_languagecountry[df_languagecountry['LanguageHaveWorkedWith'].isin(language5)]
country_language = pd.crosstab(df_languagecountry['Country'] , df_languagecountry['LanguageHaveWorkedWith'])

country_language.plot(kind = 'bar' , figsize = (12 , 6))
plt.xlabel('Country')
plt.ylabel('Count')
plt.title('Top Languages by Country')
plt.xticks(rotation=45)
plt.show()
```
