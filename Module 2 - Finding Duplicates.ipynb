## Identify and Analyze Duplicates


### Task 1: Identify Duplicate Rows
1. Count the number of duplicate rows in the dataset.
3. Display the first few duplicate rows to understand their structure.


```python
## Write your code here
duplicate_count = df.duplicated().sum()
df_duplicates = df[df.duplicated()]
print('number of duplicate rows:' , duplicate_count)
print('first few duplicate rows:')
df_duplicates
```

### Task 2: Analysis of Duplicate Response Patterns
1. Identify duplicate response patterns based on selected columns such as MainBranch, Employment, and RemoteWork.
2. Clarify that these represent multiple respondents with identical answers rather than duplicate records. Analyse which other columns frequently share identical values within these response-pattern groups.
   


```python
# We have already confirmed that there are 20 fully duplicate rows within the data set earlier.
#almost every single response in the df_mer subset, however, has a duplicate response.
df_mer = df[df.duplicated(subset = ['MainBranch' , 'Employment' , 'RemoteWork'] , keep = False)]

#If two columns share the same ResponseId, it's incredibly likely that these are fully duplicate rows. 
#So this is the column we will use to check and filter out a duplicate response.

print('duplicate ids:' , df_mer['ResponseId'].duplicated().sum())

#12 responses share an ID, now we can double check by seeing if the entire rows are duplicates as well.
print('duplicate rows:' , df_mer.duplicated().sum())

#the number of duplicate ids and fully duplicate rows are the same number, confirming that a duplicate id is a strong indicator of a duplicate record.
#now we need to remove the confirmed duplicate records

df_mer = df_mer.drop_duplicates(subset = ['ResponseId'])

#check duplicates are removed

df_mer[df_mer.duplicated()]
```

### Task 3: Visualize Shared Response Patterns
1. Create visualizations to show the distribution of shared response patterns across different categories.
2. Use bar charts or pie charts to represent the distribution of respondents who share identical values for MainBranch, Employment, and RemoteWork, grouped by Country and Employment.


```python
## Write your code here
c_count = df_mer['Country'].value_counts().head(10)

plt.figure(figsize=(10,5))

plt.bar(c_count.index, c_count.values)

plt.title('Top 10 Countries with Shared Response Patterns')
plt.xlabel('Country')
plt.ylabel('Number of Respondents')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
e_count = df_mer['Employment'].value_counts().head(10)

plt.figure(figsize=(10,5))

plt.bar(e_count.index, e_count.values)

plt.title('Top 10 Employment status with Shared Response Patterns')
plt.xlabel('Employment')
plt.ylabel('Number of Respondents')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
