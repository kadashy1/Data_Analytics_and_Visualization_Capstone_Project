The functions below will download the dataset into your browser:



```python
# Define the URL of the dataset
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### 2. Explore the Dataset
##### Task 1: Display basic information and summary statistics of the dataset.


```python
## Write your code here
df.describe()
```

### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
## Write your code here
pd.set_option('display.max_rows' , None)
pd.set_option('display.max_columns' , None)
missing_values = df.isna().sum()
missing_values = missing_values[missing_values > 0]
order_missing_values = missing_values.sort_values(ascending = True).head(5)

order_missing_values
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
## Write your code here
bottom5_missing = missing_values.sort_values(ascending = True).head(5).index

plt.figure(figsize = (12 , 6))

sns.heatmap(df[bottom5_missing].isna() , cbar = False , yticklabels = False , cmap = ['white' , 'red'])

plt.title('bottom 5 missing values heatmap')
plt.xlabel('Columns')
plt.show()
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).


```python
## Write your code here
df['CodingActivities'].isna().sum()
```

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


```python
## Write your code here
CodingActivities_mode = df['CodingActivities'].mode()[0]
CodingActivities_mode
```

##### Task 6: Impute missing values in the `Employment` column with the most frequent value.



```python
## there are no missing values in the employment column so I'm just going to keep doing the CodingActivities column...

print('missing values in CodingActivities:' , df['CodingActivities'].isna().sum())
df['CodingActivities'] = df['CodingActivities'].fillna(CodingActivities_mode)
print('missing values left in CodingActivities:' , df['CodingActivities'].isna().sum())
```

### 5. Visualizing Imputed Data
##### Task 7: Visualize the distribution of a column after imputation (e.g., `Employment`).


```python
## Write your code here
top5_CodingActivities = df['CodingActivities'].value_counts().head(5)

plt.figure(figsize = (25 , 20))

sns.barplot(x = top5_CodingActivities.index , y = top5_CodingActivities.values)

plt.title('5 most popular ways people code')
plt.xlabel('Coding Activity')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
